3. Faça um algoritmo para ler dois números e imprimir o maior, o menor ou então dizer se são iguais.

1 - Entrada: Lê dois números (num1 e num2).

2 - Decisão / Teste Condicional:

Condição 1: Se num1 > num2, então num1 é o maior e num2 é o menor.

Condição 2: Senão, se num2 > num1, então num2 é o maior e num1 é o menor.

Caso Contrário: Se nenhuma das anteriores for verdadeira, significa que os dois números são exatamente iguais.

3 - Saída: Exibe o resultado correspondente à verificação.

```python
num1 = float(input("informe o 1 número "));
num2 = float(input("informe o 2 número "));

if num1 > num2:
  print(f"O {num1} é maior");
elif num1 < num2:
  print(f"o {num1} é menor");
else:
  print(f"O resultado é igual");
```

8. Faça um algoritmo para ler três números e imprimir se estes podem ou não formar um triângulo.

1 - Entrada de Dados: Lê os três números informados pelo usuário.

2 - Validação: Aplica a regra da desigualdade triangular, onde a soma de quaisquer dois lados precisa ser obrigatoriamente maior que o terceiro lado.

3 - Decisão:

Se a condição for verdadeira para todas as 3 combinações de pares, exibe que é possível formar o triângulo.

Caso contrário, exibe que não é possível formar.

```python
num1 = int(input("informe o 1 número "));
num2 = int(input("informe o 2 número "));
num3 = int(input("informe o 3 número "));

if num1 + num2 > num3 and num2 + num3 > num1 and num1 + num3 > num2:
  print(f"O triangulo foi feito");
else:
  print(f"O triangulo não foi formado");
```

9. Faça um algoritmo para ler três números e se estes poderem formar um triângulo dizer se o triângulo é “EQUILÁTERO”, “ISÓSCELES” OU “ESCALENO”.

1 - Entrada de Dados: Lê os valores referentes aos três lados.

2 - Validação (Condição de Existência): Verifica se a soma de quaisquer dois lados é estritamente maior que o terceiro lado.

3 - Se não for possível formar: Exibe a mensagem "Não forma um triângulo" e encerra.

4 - Classificação (se formar o triângulo):

Equilátero: Verifica se todos os três lados são iguais (lado1 = lado2 e lado2 = lado3).

Escaleno: Verifica se todos os três lados são diferentes entre si (lado1 ≠ lado2, lado1 ≠ lado3 e lado2 ≠ lado3).

Isósceles: Caso não seja equilátero nem escaleno, possui exatamente dois lados iguais.

Saída: Exibe o tipo do triângulo correspondente.

```python
num1 = int(input("informe o 1 número "));
num2 = int(input("informe o 2 número "));
num3 = int(input("informe o 3 número "));

if num1 + num2 > num3 and num2 + num3 > num1 and num1 + num3 > num2:
  print(f"O triangulo foi formado");
  if num1 == num2 and num3 == num2 and num3 == num1:
    print(f"O triangulo equilatero foi feito");
  elif num1 != num2 and num3 != num2 and num3 != num1:
    print(f"O triangulo escaleno foi feito");
  elif num1 == num2 or num1 == num3 or num2 == num3:
    print(f"O triangulo isoseles foi feito");
else:
  print(f"O triangulo não foi formado");
```

10. Faça um algoritmo para ler o preço de compra de uma mercadoria e calcular o seu preço de venda para que possa ser obtido um lucro de 30%.

1 - Entrada: Solicita e armazena o preço de compra do produto.

2 - Cálculo do Lucro: Multiplica o preço de compra por 0.30 para obter o valor correspondente a 30% de margem.

3 - Cálculo da Venda: Soma o valor do lucro ao preço original de compra (ou diretamente multiplica o preço de compra por 1.30).

11. Faça um algoritmo para ler os catetos de um triângulo retângulo e calcular e imprimir a sua hipotenusa.

1 - Entrada 1: Solicita e lê o valor do primeiro cateto.

2 - Entrada 2: Solicita e lê o valor do segundo cateto.

3 - Cálculo: Eleva o primeiro cateto ao quadrado, eleva o segundo cateto ao quadrado, soma os dois resultados e extrai a raiz quadrada dessa soma.

4 - Saída: Exibe o valor obtido para a hipotenusa.

```python
num1 = int(input("informe o 1 número "));
num2 = int(input("informe o 2 número "));

hipotenusa = (num1 ** 2 + num2 ** 2) ** 0.5;

print(f"Valor da hipotenus é {hipotenusa}");
```

12. Faça um algoritmo para ler o raio e calcular o comprimento, a área e o volume de uma esfera.

1 - Entrada: Lê o valor do raio digitado pelo usuário

2 - Processamento:
Calcula o comprimento multiplicando 2 * pi * raio.
Calcula a área multiplicando 4 * pi * raio².
Calcula a volume multiplicando 4/3 * pi * raio³.

3 - Saída: Exibe os resultados do comprimento, área e volume.Fim: Encerra o programa.

```python
raio = float(input("informe o 1 número "));

pi = 3.14;
comprimento = 2 * pi * raio;
area = 4 * pi * raio ** 2;
volume = 4/3 * pi * raio ** 3;

print(f"Comprimento {comprimento}");
print(f"Área {area}");
print(f"Volume {volume}");
```
